PI Calculation using Monte Carlo Method

```python
import numpy as np

def monte_carlo(n, seed=None):
    if seed is not None:
        np.random.seed(seed)
    pts = np.random.random((n, 2))  # points in [0,1]x[0,1]
    inside = np.sum(((pts - 0.5) ** 2).sum(axis=1) <= 0.5 ** 2)
    return 4 * inside / n

# run simulation
n = 100000
pi_sim = monte_carlo(n, seed=42)

print(f"simulated pi ({n} pts): {pi_sim}")
```

```python
import random
r = 0 
N = 100000 
for i in range(N):
    x = random.uniform(0, 1)
    y = random.uniform(0, 1)
    if x**2 + y**2 <= 1:
        r += 1
pi_estimate = 4 * r / N
print(f"estimated pi ({N} pts): {pi_estimate}")
```

```python
import random
import matplotlib.pyplot as plt

# Parameters
N = 10000 # number of points
inside_x = []
inside_y = []
outside_x = []
outside_y = []

# Simulation
r = 0
for _ in range(N):
    x = random.uniform(0, 1)
    y = random.uniform(0, 1)
    if x**2 + y**2 <= 1:
        r += 1
        inside_x.append(x)
        inside_y.append(y)
    else:
        outside_x.append(x)
        outside_y.append(y)

# Estimate π
pi_estimate = 4 * r / N
print(f"Estimated π ({N} points): {pi_estimate}")

# Visualization
plt.figure(figsize=(6,6))
plt.scatter(inside_x, inside_y, color='blue', s=2, label='Inside Circle')
plt.scatter(outside_x, outside_y, color='red', s=2, label='Outside Circle')
plt.title(f"Monte Carlo π Approximation (Estimated π = {pi_estimate:.5f})")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.axis('square')
plt.show()
```

```python
# Vectorization using NumPy
import numpy as np
N = 100000
points = np.random.rand(N, 2)  # Generate N random points in [0,1]x[0,1]
distances = np.sum(points**2, axis=1)  # Calculate distance from origin
inside_count = np.sum(distances <= 1)
pi_estimate = 4 * inside_count / N
print(f"Estimated π using NumPy ({N} points): {pi_estimate}")   
```

```python
# Using Numba for nJIT compilation
from numba import njit
import numpy as np      
@njit
def monte_carlo_numba(n):
    r = 0
    for _ in range(n):
        x = np.random.rand()
        y = np.random.rand()
        if x**2 + y**2 <= 1:
            r += 1
    return 4 * r / n
N = 100000
pi_estimate_numba = monte_carlo_numba(N)
print(f"Estimated π using Numba ({N} points): {pi_estimate_numba}")
```

```python
from dask import array as da
import numpy as np

x = np.random.rand(10000, 10000)
dx = da.from_array(x, chunks=(250, 250))

print(dx)
print(dx.mean().compute())
print(dx.chunks)
```

```python
dx2 = da.random.random((10000, 10000), chunks=(250, 250))
print(dx2)
```

```python
# inspect chunk layout
dx = da.random.random((10000.0, 10000.0), chunks='auto')
print(dx.chunks)
```


Exercise 4.1: "Pi calculation using Monte Carlo method"
Implement the calculation of the irrational number Pi using the Monte Carlo method
Use Python multiprocessing to gain access to multiple processes
Show the result for increasing number of random realizations 
Measure the execution time and observe the speedup when increasing the number of processors

```python
import multiprocessing
import random
import time

def monte_carlo_pi_part(num_points):
    """Calculate number of points inside the unit circle for a chunk of points."""
    count = 0
    for _ in range(num_points):
        x, y = random.random(), random.random()
        if x*x + y*y <= 1.0:
            count += 1
    return count

def estimate_pi(total_points, num_processes):
    """Estimate Pi using multiprocessing."""
    # Divide total points among processes
    points_per_process = total_points // num_processes
    with multiprocessing.Pool(processes=num_processes) as pool:
        results = pool.map(monte_carlo_pi_part, [points_per_process]*num_processes)
    
    # Total points inside circle
    total_inside_circle = sum(results)
    
    # Monte Carlo estimation of Pi
    pi_estimate = 4 * total_inside_circle / total_points
    return pi_estimate

if __name__ == "__main__":
    total_points = 10_000_000  # Total random points
    process_counts = [1, 2, 4, 8]  # Different numbers of processes

    print(f"Estimating Pi using {total_points} random points\n")
    
    timings = []

    for num_proc in process_counts:
        start_time = time.time()
        pi_value = estimate_pi(total_points, num_proc)
        end_time = time.time()
        elapsed_time = end_time - start_time
        timings.append(elapsed_time)
        print(f"Processes: {num_proc}, Pi ≈ {pi_value:.6f}, Time: {elapsed_time:.2f}s")
    
    # Compute speedup
    print("\nSpeedup relative to single process:")
    for i, num_proc in enumerate(process_counts):
        speedup = timings[0] / timings[i]
        print(f"Processes: {num_proc}, Speedup: {speedup:.2f}x")
```
